How to remove duplicates from a string?

```python
string = 'programming'
result = " "

for char in string :
    if char not in result:
         result += char
    print(result)
```

Count frequency of elements

```python
numbers=[1,2,2,3,3,3]
freq = {}
for num in numbers:
      freq[num] = freq.get(1,0)+1
print(freq)
```

Remove Duplicate Elements from a List in Python

```python
number = [1,2,2,2,1,1,1,3,4,5,6,7,8,12,13,12,14,12,13,14,15]
result = list(set(number))
print(result)
```

Remove Duplicate Elements from a list Without using set 

```python
numbers = [1,2,2,2,1,1,1,3,4,5,6,7,8,12,13,12,14,12,13,14,15]
result = []

for num in numbers:
    if num not in result:
       result.append(num)

print(result)


    
```

Remove Duplicate Elements from a list Using Dictionary (Maintains Order)

```python
numbers = [1,2,2,2,1,1,1,3,4,5,6,7,8,12,13,12,14,12,13,14,15]
result = list(dict.fromkeys(numbers))
print(result)
```
