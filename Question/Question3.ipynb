Check if a string is palindrom or not 

```python
def is_palindrome(str):
    len_str = len(str)
    c = 0
    for i in range(len_str//2):
        if str[i] == str[len_str -i-1]:
            continue
        else:
            c = 1
            break
    return c == 0
    print(is_palindrome(str))
```

```python
text = "saziya"
if text == text[::-1]:
    print("palindrome")
else:
    print("not palindrome")
```
