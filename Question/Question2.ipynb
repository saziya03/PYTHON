Find a non - negative integer that is not in the array

```python
l = [0,1,2,3,4,5,6,7,9]
def find_missing_number(l):
  list_dict = {}
  for i in l:
    list_dict[i] = True

  for i in range(len(l)+1):
    print(i, list_dict.get(i))
    if not list_dict.get(i):
      print("number",i)
      return i
  return None

print(find_missing_number(l))
```
